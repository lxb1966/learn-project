```python
#函数
def min_max_normalize(x):
    x=np.array(x,dtype=float)#x为向量，dtype使其变成浮点数
    return(x-x.min())/(x.max()-x.min())
values=[80,85,90,88,100]
normalized_values=min_max_normalize(values)
print(normalized_values)
```

```python
#学习numpy
x=np.array([80,87,54,42,45,90])#列表里必须数字
print(f"平均值:{np.mean(x)}")#np.mean()
print(f"中位数:{np.median(x)}")#np.median()
print(f"标准差:{np.std(x)}")#np.std()
#平均值还可以直接算
print("平均值为",x.mean())#这里不需要np.
#np.std()标准差
#np.max()与np.min()为最大最小值
```

```python
#均值与标准差
#手动计算均值
manual_mean=np.sum(x)/len(x)#len()为自带函数，计算长度
print(f"手动计算的平均值:{manual_mean}")
print("numpy计算均值:",x.mean())
```

```python
# 示例数据：7 个城市的 GDP、人口、绿化率（数值为示例，仅用于练习）
city_data = pd.DataFrame({
    "城市": ["北京", "上海", "深圳", "广州", "成都", "杭州", "武汉"],
    "GDP_亿元": [43760.7, 47218.7, 34606.4, 30355.7, 22074.7, 20059.0, 20011.7],
    "人口_万人": [2185.8, 2487.5, 1779.0, 1882.7, 2140.3, 1252.2, 1377.4],
    "绿化率_百分比": [49.8, 39.5, 45.2, 43.1, 42.6, 40.8, 39.6],
})

df_city = pd.DataFrame(city_data)
# 亿元 / 万人 = 万元/人，加一列人均 GDP 方便后续分析
df_city["人均GDP_万元"] = (df_city["GDP_亿元"] / df_city["人口_万人"]).round(2)

print("数据规模：", df_city.shape)          # (行数, 列数)
print("列名：", list(df_city.columns))
df_city
```
